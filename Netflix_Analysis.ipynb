```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("/content/netflix_titles.csv")
```

```python
df.head()
```

```python
df.info()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
df["director"] = df["director"].fillna("Unknown")
df["cast"] = df["cast"].fillna("Unknown")
df["country"] = df["country"].fillna("Unknown")
```

```python
df["rating"] = df["rating"].fillna(df["rating"].mode()[0])
df["duration"] = df["duration"].fillna(df["duration"].mode()[0])
df["date_added"] = df["date_added"].fillna(df["date_added"].mode()[0])
```

```python
# Count total Movies and TV Shows.
df["type"].value_counts().plot(kind = "pie")
plt.title("Movie VS TV Show")
plt.show()
```

```python
df.head()
```

```python
# Find the top 5 countries with the most Netflix content.
df["country"].value_counts().head(10).plot(kind = "bar")
plt.title("Top 10 Country")
plt.show()
```

```python
# Show all TV Shows released after 2018.
df[(df["type"] == "TV Show") & (df["release_year"] > 2018)]
```

```python
# Find all content with rating "TV-MA".
df[df["rating"] == "TV-MA"]
```

```python
# Count how many Movies were released in 2020.
len(df[(df["type"] == "Movie") & (df["release_year"] == 2020)])
```

```python
# Find titles added in 2021.
df[df["date_added"].str.contains("2021")]
```

```python
# Find the oldest movie/show in the dataset.
oldest = df["release_year"].min()
result = df[df["release_year"] == oldest]
print(result["title"])
```

```python
# Find the newest release year available.
df["release_year"].max()
```

```python
# Show all content from India.
df[df["country"] == "India"]
```

```python
# Find how many unique ratings exist.
df['rating'].nunique()
```

```python
# Find the most common rating.
df['rating'].mode()
```

```python
# Show all titles directed by "Rajiv Chilaka".
df[df["director"] == "Rajiv Chilaka"]
```

```python
# Find all content where cast contains "Shah Rukh Khan".
df[df["cast"].str.contains("Shah Rukh Khan")]
```

```python
# Find all Movies from India released after 2015.
df[(df["type"] == "Movie") & (df["country"] == "India") & (df["release_year"] > 2015)]
```

```python
# Count how many titles belong to each rating category.
df.groupby("rating").count()["title"]
```

```python
# Find top 10 release years with most content.
df["release_year"].value_counts().head(10)
```

```python
# Find the average release year of Movies and TV Shows separately.
df.groupby("type")["release_year"].mean()
```

```python
# Create pie chart for ratings distribution.
df["rating"].value_counts().plot(kind = "pie")
plt.title("Rating Distribution")
plt.show()
```

```python
# Create histogram for release years.
df["release_year"].value_counts().plot(kind = "hist")
plt.title("Release Year Distribution")
plt.show()
```

```python
# Find all titles where country contains "United States".
result = df[df["country"].str.contains("United States")]
print(result["title"])

```

```python
# Count how many titles were added each year.
df["year_added"] = df["date_added"].str[-4:]
print(df["year_added"].value_counts())
```
